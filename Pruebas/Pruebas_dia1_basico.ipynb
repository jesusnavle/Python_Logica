Entrada: nums = [2, 7, 11, 15], objetivo = 9

Salida: [0, 1]

```python
nums = [2, 7, 11, 15]
objetivo = 9

for i in nums:
    for j in nums:
        if i + j == 9:
         print(i, j)
         print(nums.index(i))
         print(nums.index(j))
```

```python
nums = [2, 7, 11, 15]
objetivo = 9

def sum_pa(nums, objetivo):
    salida = []
    for i in nums:
        for j in nums:
                if i + j == 9:
                    i = nums.index(i)
                    j = nums.index(j)
                    salida.append(i)
                    
    return salida

print(sum_pa(nums, objetivo))
```

Palíndromo de Números: Crea un programa que determine si un número entero es un palíndromo (se lee igual de izquierda a derecha que de derecha a izquierda) sin convertir el número a string.

Entrada: 121 -> Salida: True

Entrada: -121 -> Salida: False

```python
numero = 1211
original = numero
invertido = 0

while numero > 0:
    dijito = numero % 10
    invertido = invertido * 10 + dijito
    numero //= 10

invertido

if original == invertido:
    print(f"El número es un palindromo {True}")
else:
    print(f"El número no es un palindromo {False}")
    
```

Contador de Frecuencias: Dada una cadena de texto, cuenta cuántas veces aparece cada carácter y devuélvelo en un diccionario.

Entrada: "programacion"

Salida: {'p': 1, 'r': 2, 'o': 2, 'g': 1, 'a': 2, 'm': 1, 'c': 1, 'i': 1, 'n': 1}

```python
entrada = 'programacion'
salida = {}

for letra in entrada:
    if letra in salida:
        salida[letra] += 1
    else:
        salida[letra] = 1

print(salida)


#PRACTICAR MAS DICCIONARIOS.
```

Eliminar Duplicados: Dada una lista con elementos repetidos, crea una función que devuelva una nueva lista con los elementos únicos, manteniendo el orden original.

Entrada: [1, 2, 2, 3, 4, 4, 5]

Salida: [1, 2, 3, 4, 5]

```python
entrada = [1, 2, 2, 3, 4, 4, 5]
salida = []

for i in entrada:
    if i not in salida:
        salida.append(i)

print(salida)


#practicar mas listas
        
```
