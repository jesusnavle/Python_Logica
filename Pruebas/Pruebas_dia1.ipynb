```python
#1. Nivel Básico: Manipulación de Estructuras (El "FizzBuzz")
#Reto: Escribe una función que reciba una lista de números y devuelva una nueva lista con los cuadrados de los números, 
# pero solo de aquellos que sean pares.

#Pista: En Python, lo ideal es usar List Comprehensions en lugar de un ciclo for tradicional con un .append().
```

```python
import random

lista = []

for _ in range(10):
    lista.append(random.randint(1, 24))
print(f"lista principal {lista}")

def numeros_cuadrado(lista):

    lista_nueva = []

    for numero in lista:
        if numero % 2 == 0:
            cuadrado = numero ** 2
            lista_nueva.append(cuadrado)

    return lista_nueva

print("la lista cuadrada es:")
print(numeros_cuadrado(lista))
```
