1. El Portero de la Discoteca (Filtro de Listas)
Este ejercicio te ayudará a practicar el recorrido de listas y la creación de nuevas listas basadas en una condición.

Reto: Tienes una lista de edades. Crea una nueva lista que solo contenga las edades que son mayores o iguales a 18.

Entrada: edades = [12, 25, 17, 18, 30, 15]

Salida esperada: [25, 18, 30]

```python
edad = [12 , 25, 18, 30, 15]
edad_18 = []

for i in edad:
    if i >= 18:
        edad_18.append(i)

print(f'las edades mayores a 18 son: {edad_18}')
```

2. El Buscador de Índices (Método enumerate)
Para el ejercicio de "Suma de Pares" (Target Sum), necesitas saber en qué posición están los números. Este ejercicio te enseña a usar enumerate().

Reto: Dada una lista de nombres, imprime el nombre seguido de su posición (índice) solo si el nombre empieza con la letra "A".

Entrada: nombres = ["Ana", "Pedro", "Alberto", "Juan"]

Salida esperada:
Ana está en el índice 0
Alberto está en el índice 2

```python
nombres = ["Ana", "Pedro", "Alberto", "Juan"]
inicial = "A"

for indice, letra in enumerate(nombres):
    if inicial == letra[0]:
        print(str(indice) + letra)
```
