El Sumador de Largos
Dada una lista de palabras, suma la longitud de todas ellas.

Entrada: ["hola", "sol"]

Salida esperada: 7 (porque 4 + 3 = 7).

```python
palabras = ["hola", "sol"]

sum_palabras = len(palabras[0]) + len(palabras[1])
print(sum_palabras)
```

```python
#Corrección

palabras = ["hola", "sol"]
sum = 0

for i in palabras:
    sum += len(i)

print(sum)
```

2. Multiplicar solo los Positivos
Tienes una lista de números. Crea una nueva lista donde multipliques por 10 solo a los números que sean positivos. Los negativos ignóralos.

Entrada: [1, -2, 3] -> Salida esperada: [10, 30]

```python
list_numbers = [1, -2, 3]
list_mult = []

for i in list_numbers:
    if i >= 0:
        list_mult.append(i * 10)

print(list_mult)
        
```

Buscador de Letra "A"
Recorre una lista de nombres. Si el nombre comienza con la letra "A", imprime "Nombre válido".

Pista: Puedes usar if nombre[0] == "A":.

Entrada: ["Ana", "Saul", "Alberto"]

```python
names = ["Ana", "Saul", "Alberto"]
names_A = []

for nam in names:
    if nam[0] == "A":
        names_A.append(nam)

print(names_A)
```

Reemplazo de Elementos
Dada una lista de números, recórrela. Si encuentras un 0, imprime "Cero detectado". Si encuentras cualquier otro número, imprime "Número normal".

Entrada: [5, 0, 2, 0]

```python
numbers = [5, 0, 2, 0]

for i in numbers:
    if i == 0:
        print(f"{i} Cero detectado")
    else:
        print(f"{i} Número normal")
```

Contador de Palabras Largas vs Cortas
Dada una lista de palabras, cuenta cuántas tienen más de 4 letras y cuántas tienen 4 o menos. Imprime ambos totales.

Entrada: ["casa", "elefante", "ojo"]

Salida esperada: Largas: 1, Cortas: 2

```python
palabras = ["casa", "elefante", "ojo"]

cont_largo = 0
cont_corto = 0

for i in palabras:
    if len(i) > 4:
        cont_largo += 1
    else:
        cont_corto += 1

print(cont_largo)
print(cont_corto)

print(f"Palabras Largas: {cont_largo}, Palabras Cortas: {cont_corto}")
```
