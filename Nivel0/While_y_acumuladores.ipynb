1. La Cuenta Regresiva (while)
Usa un ciclo while para imprimir los números del 10 al 1. Al final, imprime "¡Despegue!".

Pista: Empieza con numero = 10 y ve restando 1 en cada vuelta.

```python
x = 11
while x > 1:
    x -= 1
    print(x)
print("¡Despegue!")
```

Escribe un programa que pida números al usuario y los vaya sumando. El ciclo debe detenerse solo cuando la suma total sea mayor o igual a 50.

Pista: Necesitas una variable total = 0.

3. El Promediador de Calificaciones
Tienes una lista de calificaciones: [80, 90, 70, 100]. Calcula el promedio sumándolas todas con un for y dividiendo entre la cantidad de elementos.

Entrada: [80, 90, 70, 100] -> Salida esperada: 85.0

```python
calificaciones = [80, 90, 70, 100]
suma_calificaciones = 0

for i in calificaciones:
    suma_calificaciones += i

promedio = suma_calificaciones / len(calificaciones)

print("el promedio de tus calificaciones es: ", promedio)
```

5. Tabla de Multiplicar (for)
Pide un número al usuario e imprime su tabla de multiplicar del 1 al 10.

Ejemplo: Si el usuario pone 5, debe salir 5 x 1 = 5, 5 x 2 = 10... hasta el 10.

```python
tabla = int(input("Escribe un número para la tabla de multiplicar: "))

print(f"El número que elejiste es {tabla}")

for i in range(1, 11):
    resultado = tabla * i
    print(f"{tabla} * {i} = {resultado}")
```

```python
#Codigo con while

#utilizamos el -1 para iniciar el programa sin que se cierre
numero = -1

#Iniciamos el ciclo while !0 para indicar que cuando sea = 0 se cierre el programa
while numero != 0:
    numero = int(input("Escribe el número a multiplicar: "))
    if numero != 0:
        for i in range(1, 11):
            print(f"{numero} x {i} = {numero * i}")
```
