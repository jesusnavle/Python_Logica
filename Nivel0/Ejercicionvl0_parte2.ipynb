1. El Cuenta-Vocales
Este ejercicio es un clásico. Te ayudará a comparar un elemento contra varias opciones.

Reto: Dada una palabra (ej. "computadora"), recorre cada letra y cuenta cuántas vocales tiene en total.

Pista: Puedes usar if letra in "aeiou":.

Entrada: "programacion" -> Salida esperada: 5

```python
letra = "computadora"
cont_letra = 0

for i in letra:
    if i in "aeiou":
        cont_letra += 1

print(cont_letra)
```

2. Multiplicador de Lista
Aquí practicarás la creación de nuevas listas a partir de una existente.

Reto: Tienes una lista de números. Crea una nueva lista donde cada número de la original esté multiplicado por 2.

Entrada: [1, 2, 3] -> Salida esperada: [2, 4, 6]

```python
num = [1, 2, 3]
num_mult = []

for i in num:
    num_mult.append(i * 2)

print(num)
print(num_mult)
    
```

3. Buscador de "Software"
A veces necesitamos buscar una palabra específica dentro de una lista de frases.

Reto: Tienes una lista de oraciones. Recórrela e imprime solo las oraciones que contengan la palabra "Python".

Entrada: ["Me gusta Java", "Python es genial", "C++ es dificil", "Aprendo Python"]

```python
frases = ["Me gusta Java", "Python es genial", "C++ es dificiel", "Aprendiendo Python"]

for palabra in frases:
    if palabra == "Python":
        print(palabra)

print(palabra)
    
```

```python
#Corregido 

frases = ["Me gusta Java", "Python es genial", "C++ es dificil", "Aprendiendo Python"]

for frase in frases:
    if "Python" in frase:
        print(frase)
```

4. ¿Es el número 7?
Este te ayudará a usar el if para encontrar algo y detenerte si quieres (aunque por ahora solo imprímelo).

Reto: Recorre una lista de números. Si el número es 7, imprime "¡Encontré el siete!". Si no es 7, imprime "No es el siete".

Entrada: [1, 7, 3, 7]

```python
numero = [1, 7, 3, 7]
#numero = [1, 8, 3, 8]


for i in numero:
    if i == 7:
        print("Encontre el número 7")
    else:
        print("No encontre el número 7")
```

5. Filtrado de Textos Cortos
Para practicar la función len(), que mide el tamaño de las cosas.

Reto: Dada una lista de palabras, imprime solo aquellas que tengan más de 5 letras.

Entrada: ["sol", "estrella", "luz", "universo"]

Salida esperada:

```python
letra = ["sol", "estrella", "luz", "universo"]

for i in letra:
    if len(i) >= 5:
        print(i)
```
