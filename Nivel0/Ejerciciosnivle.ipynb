Saludo Selectivo: Crea una lista con los nombres "Saúl", "Diego" y "Ana". Haz un ciclo que recorra la lista, pero que solo imprima un saludo especial si el nombre es "Saúl".

```python
nombre = ['Saul', 'Diego', 'Ana']

for i in nombre:
    if i == 'Saul':
        print(i)
```

Contador de Positivos: Tienes la lista valores = [10, -5, 8, -2, 1]. Crea un ciclo que sume solo los números que sean mayores a 0.

```python
valores = [10, -5, 8, -2, 1]
suma = 0

for positivos in valores:
    if positivos >= 0:
        suma += positivos 

print(suma)
```

El Grito: Recorre una palabra (por ejemplo "python") y usa un ciclo para imprimir cada letra en mayúsculas (puedes usar .upper()).

```python
letra = "python"
for mayuscula in letra.upper():
    print(mayuscula)
```
