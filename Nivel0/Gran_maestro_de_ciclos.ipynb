1. El Validador de Contraseña (while)
Crea un programa que pida una contraseña. Mientras la contraseña no sea "python123", debe decir "Acceso denegado" y volver a pedirla. Cuando sea correcta, imprime "Bienvenido al sistema".

```python
password = False

while password == False:
    contraseña = input("Introduce la contraseña: ")
    if contraseña == "python123":
        print("Bienvenido a Sistema")
        break
    else:
        print("Acceso denegado :c")
```

2. El Contador de "Pan" (for + contador)
Tienes una lista de compras: ["leche", "pan", "huevo", "pan", "queso", "pan"]. Crea un ciclo que cuente cuántas veces aparece la palabra "pan" y al final imprima el total.

Entrada: La lista mencionada.

Salida esperada: El pan aparece 3 veces.

```python
compras = ["leche", "pan", "huevo", "pan", "queso","pan"]
contador = 0

for i in compras:
    if i == "pan":
        contador += 1
        
print(f"El {i} aparece {contador} veces")
```

```python
#prueba con ciclo while y pidiendo a usuario

inicio = False
compras = ["leche", "leche", "pan", "huevo", "pan", "queso","pan", "queso"]
contador = 0


while inicio == False:
    print(f"Los productos que quieres contar son:")
    print(compras)
    objetivo = input("Escribe la compra que quieres contar: ")
    for i in compras:
        if i == objetivo:
            contador += 1  
    print(f"El {objetivo} aparece {contador} vece/s")
else: 
    print("producto no encotrado")
```

3. Suma Selectiva (for + if)
Dada una lista de números [10, 5, 20, 3, 40], suma solo los números que sean mayores a 15.

Salida esperada: 60 (porque sumó 20 + 40).

```python
numeros = [10, 5, 20, 3, 40]
suma = 0

for i in numeros:
    if i > 15:
        print(i)
        suma += i

print(f"la suma de los números mayores a 15 son: {suma}")
```

4. Encuentra el Primer Negativo (break)
Dada la lista [5, 8, 12, -3, 7, -2], recorre la lista y, en cuanto encuentres el primer número negativo, imprímelo y detén el ciclo con break.

Salida esperada: El primer negativo es -3.

```python
numeros = [5, 8, 12, -3, 7, -2]

for i in numeros:
    if i <= 0:
        negativo = i
        print(f"el número negativo es: {negativo}")
        break
        
```

5. Adivina el Número (El Mini-Juego)
Define una variable secreto = 7. Pide al usuario que adivine el número.

Si el usuario escribe un número menor al secreto, dile "Es más alto".

Si escribe un número mayor, dile "Es más bajo".

El ciclo while debe seguir hasta que el usuario acierte.

```python
objetivo = 7
inicio = False

while inicio == False:
    seleccion = int(input("Escribe el número: "))
    if seleccion == objetivo:
        print("Acertaste")
        break
    elif seleccion > 7:
        print(f"Es mas bajo, el numero que elijiste fue: {seleccion}")
    elif seleccion < 7:
        print(f"Es mas alto, el número que elijiste fue: {seleccion}")
    
```
