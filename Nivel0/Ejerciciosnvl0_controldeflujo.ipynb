```python
lista = [1, 5, 8, 10, 15]
secreto = 10
encontrado = False  # Esta es nuestra "Bandera"

# Usamos un for para revisar la lista
for num in lista:
    if num == secreto:
        encontrado = True
        break  # Detenemos el ciclo porque ya lo encontramos

# Fuera del ciclo, revisamos cómo quedó la bandera
if encontrado:
    print("¡El número 10 está en la lista!")
else:
    print("No se encontró el número.")
```

Ahora aqui la lógica con el ciclo While
1. agregamoso la variable en tipo False para que una vez que se logre el objetivo pare el programa.

3. dentro del while agregamos una variable que le diga al usuario insertar un número para versi el número que esta pidiendo el programa o no

```python
acerto = False

while acerto == False:
    numero = int(input("Adivina el número secreto: "))
    if numero == 10:
        print("¡Ganaste!")
        break #el brake es para terminar el programa
    else:
        print("Intenta de nuevo.")
```

El Promediador: Suma todos los números de una lista y, al final, divide el total entre la cantidad de elementos (usa len(lista)) para obtener el promedio.

```python
numeros = [9, 9, 9, 9, 9, 9, 9]
suma = 0

for i in numeros:
    suma += i

promedio = suma / len(numeros)

print(promedio)
```

Filtro de nombres cortos: Crea una lista de nombres y genera una nueva lista que solo contenga los nombres que tengan exactamente 4 letras.

```python
nombres = ["Alan", "Lusi", "Diego", "Jesús", "Lulu"]
nombres_le4 = []

for i in nombres:
    if len(i) == 4:
        nombres_le4.append(i)

print(nombres_le4)
```
